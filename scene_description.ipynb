```python
import prompt_construction

from PIL import Image
import os
from pathlib import Path
from vlmx.agent import Agent, AgentConfig
import argparse
from dotenv import load_dotenv

def describe_scene_from_image(image_path, overall_task, agent=None, prompt="scene_description_v1.txt"):
    """
    Generate a scene description from an image file.
    
    Args:
        image_path (str): Path to the .jpg image file
        overall_task (str): Description of the overall task
        agent (Agent, optional): VLM agent to use. If None, a default agent will be created.
        prompt (str, optional): Name of the prompt template file to use
        
    Returns:
        str: The generated scene description
    """
    # Validate image path
    if not os.path.exists(image_path):
        raise FileNotFoundError(f"Image file not found: {image_path}")
    
    if not image_path.lower().endswith('.jpg'):
        raise ValueError(f"Expected a .jpg file, got: {image_path}")
    
    # Load the image
    image = Image.open(image_path)
    
    # If no agent is provided, create a default one
    if agent is None:
        # You may need to adjust this based on your environment setup
        from dotenv import load_dotenv
        load_dotenv()
        api_key = os.environ.get("OPENAI_API_KEY")
        if not api_key:
            raise ValueError("No API key found. Set OPENAI_API_KEY environment variable.")
        
        cfg = AgentConfig(model_name="gpt-4-vision-preview", out_dir="temp_output", api_key=api_key)
        agent = Agent(cfg)
    
    # Generate the scene description
    scene_description = prompt_construction.scene_description(
        agent=agent,
        overall_task=overall_task,
        image=image,
        prompt=prompt
    )
    
    return scene_description

def describe_scene_from_multiple_images(agent, image_paths, overall_task, prompt="scene_description_list_v4.txt"):
    """
    Generate a scene description from multiple image files.
    
    Args:
        image_paths (list): List of paths to .jpg image files
        overall_task (str): Description of the overall task
        agent (Agent, optional): VLM agent to use. If None, a default agent will be created.
        prompt (str, optional): Name of the prompt template file to use
        
    Returns:
        str: The generated scene description
    """
    # Validate image paths
    for path in image_paths:
        if not os.path.exists(path):
            raise FileNotFoundError(f"Image file not found: {path}")
        
        if not path.lower().endswith('.jpg'):
            raise ValueError(f"Expected a .jpg file, got: {path}")
    
    # Load the images as PIL Image objects
    images = [Image.open(path).copy() for path in image_paths]
    
    # For debugging - verify we have PIL Image objects, not just paths
    print(f"Type of images list: {type(images)}")
    print(f"Type of first image: {type(images[0])}")
    
    # If we're using Gemini, we need to pass a single image, not a list
    # The error in the output suggests we need to handle this differently
    # if "gemini" in str(agent.config.model_name).lower():
    #     # For Gemini models, we might need to use the first image only
    #     # or implement a different approach for multiple images
    #     print("Using Gemini model - note that it may need special handling for multiple images")
    
    
    # Generate the scene description
    scene_description = prompt_construction.scene_description(
        agent=agent,
        overall_task=overall_task,
        image=images,
        prompt=prompt
    )
    
    return scene_description
```
